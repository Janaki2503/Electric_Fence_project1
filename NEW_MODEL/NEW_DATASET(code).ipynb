```python
import pandas as pd
import random

data = []

for i in range(3000):

    # Random PIR
    pir = random.choice([0, 1])

    # SAFE (0)
    if i < 1000:
        vibration = random.randint(110, 350)
        voltage = round(random.uniform(22.0, 24.0), 1)
        current = round(random.uniform(0.4, 0.8), 1)
        risk = 0

    # MEDIUM (1)
    elif i < 2000:
        vibration = random.randint(350, 700)
        pir = 1
        voltage = round(random.uniform(21.0, 22.5), 1)
        current = round(random.uniform(0.8, 1.2), 1)
        risk = 1

    # HIGH (2)
    else:
        vibration = random.randint(700, 2000)
        pir = 1
        voltage = round(random.uniform(17.0, 21.0), 1)
        current = round(random.uniform(1.3, 2.5), 1)
        risk = 2

    data.append([voltage, current, pir, vibration, risk])

# Create DataFrame
df = pd.DataFrame(data, columns=[
    "voltage", "current", "pir", "vibration", "risk"
])

# Save CSV
df.to_csv("fence_3level_dataset.csv", index=False)

print("✅ Dataset generated successfully (3000 rows)")
```
