```python
import joblib
import numpy as np

# 📥 Load trained model
model = joblib.load("fence_model_3level.pkl")

# 🔍 Test samples (you can change values)
test_data = [
    [23.0, 0.6, 0, 200],   # SAFE
    [21.5, 1.0, 1, 500],   # MEDIUM
    [19.0, 1.8, 1, 1200]   # HIGH
]

print("🔎 Testing Model...\n")

for i, data in enumerate(test_data):

    voltage, current, pir, vibration = data

    # Convert to numpy
    input_data = np.array([data])

    # Predict
    prediction = model.predict(input_data)[0]

    # 🔥 Extra safety (avoid false alarm)
    if vibration < 350:
        prediction = 0

    # 🎯 Output label
    if prediction == 0:
        result = "SAFE 🟢"
    elif prediction == 1:
        result = "MEDIUM 🟡"
    else:
        result = "HIGH 🔴"

    print(f"Test Case {i+1}:")
    print(f"Input → V:{voltage}, C:{current}, PIR:{pir}, Vib:{vibration}")
    print(f"Prediction → {result}\n")
```
